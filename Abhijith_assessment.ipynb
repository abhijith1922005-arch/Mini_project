```python
import numpy as np
import pandas as pd

np.random.seed(42)
n = 200

df = pd.DataFrame({
    "order_id": range(1001, 1001 + n),
    "order_date": pd.date_range("2024-01-01", periods=n, freq="D"),
    "customer_name": np.random.choice(
        ["Anita", "Ravi", "Meera", "Arjun", "Sneha", "Karthik", "Divya", "Suresh"], n),
    "city": np.random.choice(
        ["Kochi", "Chennai", "Bengaluru", "Hyderabad", "Mumbai"], n),
    "category": np.random.choice(
        ["Electronics", "Clothing", "Grocery", "Books", "Furniture"], n),
    "quantity": np.random.randint(1, 10, n),
    "unit_price": np.random.randint(100, 5000, n),
    "payment_mode": np.random.choice(
        ["UPI", "Card", "Cash", "NetBanking"], n)
})

df.to_csv("sales_data.csv", index=False)
print("File created with", len(df), "rows")
```

A1-Function with a default argument

```python
def calculate_total(quantity, unit_price, discount_percent=0):
    subtotal = quantity * unit_price
    discount = subtotal * (discount_percent / 100)
    final_amount = subtotal - discount
    return final_amount

print(calculate_total(3, 500))       # 1500
print(calculate_total(2, 1000, 10))  # 1800.0
```

# Section C — Pandas

```python
import pandas as pd

sales = pd.read_csv("sales_data.csv", parse_dates=["order_date"])
```

C1.

```python
print("First 5 rows:")
display(sales.head(5))

print("Last 3 rows:")
display(sales.tail(3))

print("Shape:", sales.shape)
print("Columns:", sales.columns.tolist())
print("Data types:")
print(sales.dtypes)

print("Numeric description:")
display(sales.describe())

print("Missing values per column:")
print(sales.isnull().sum())
```

C2


```python
sales["total_amount"] = sales["quantity"] * sales["unit_price"]
sales["month"] = sales["order_date"].dt.month_name()

display(sales.head())
```

C3

```python
# Orders in Kochi AND in the Electronics category
kochi_electronics = sales.loc[
    (sales["city"] == "Kochi") & (sales["category"] == "Electronics")
]
display(kochi_electronics)

# Orders where total_amount > 20000 OR payment_mode is Cash
high_or_cash = sales.loc[
    (sales["total_amount"] > 20000) | (sales["payment_mode"] == "Cash")
]
display(high_or_cash)
print("Matching rows:", len(high_or_cash))
```

C4


```python
print("Total sales by category:")
display(sales.groupby("category")["total_amount"].sum())

print("Average order value by city:")
display(sales.groupby("city")["total_amount"].mean())

print("Number of orders by payment mode:")
display(sales.groupby("payment_mode")["order_id"].count())

print("Top 3 customers by total spending:")
customer_spending = (
    sales.groupby("customer_name")["total_amount"]
    .sum()
    .nlargest(3)
)
display(customer_spending)
```

C5

```python
top_5_orders = sales.sort_values("total_amount", ascending=False).head(5)
display(top_5_orders)
```

C6


```python
sales_pivot = pd.pivot_table(
    sales,
    values="total_amount",
    index="city",
    columns="category",
    aggfunc="sum",
    fill_value=0
)
display(sales_pivot)

```

```python
import sqlite3

conn = sqlite3.connect("sales.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY,
    order_date TEXT,
    customer_name TEXT,
    city TEXT,
    category TEXT,
    quantity INTEGER,
    unit_price INTEGER,
    payment_mode TEXT
)
""")
conn.commit()
```

E2

```python
sales_for_sql = sales.copy()
sales_for_sql["order_date"] = sales_for_sql["order_date"].dt.strftime("%Y-%m-%d")

sales_for_sql[
    ["order_id", "order_date", "customer_name", "city", "category",
     "quantity", "unit_price", "payment_mode"]
].to_sql("orders", conn, if_exists="replace", index=False)

row_count = pd.read_sql_query("SELECT COUNT(*) AS total_rows FROM orders", conn)
display(row_count)
```

E3

```python
# 1. All orders from Mumbai
query1 = """
SELECT *
FROM orders
WHERE city = 'Mumbai'
"""
display(pd.read_sql_query(query1, conn))

# 2. Total quantity sold for each category
query2 = """
SELECT category, SUM(quantity) AS total_quantity
FROM orders
GROUP BY category
"""
display(pd.read_sql_query(query2, conn))

# 3. Cities where average unit price is more than 2500
query3 = """
SELECT city, AVG(unit_price) AS average_unit_price
FROM orders
GROUP BY city
HAVING AVG(unit_price) > 2500
"""
display(pd.read_sql_query(query3, conn))

# 4. Five most expensive orders by quantity * unit_price
query4 = """
SELECT *, quantity * unit_price AS total_amount
FROM orders
ORDER BY total_amount DESC
LIMIT 5
"""
display(pd.read_sql_query(query4, conn))

conn.close()
```

# E5. SQLite vs MySQL

SQLite is a lightweight, file-based database and does not require a separate database server. MySQL is a client–server database system designed for multi-user and larger applications. I would choose SQLite for small applications, prototypes, or local analysis; I would choose MySQL for applications that need concurrent users, centralized access, and server-based management.



#Section F: Mini ETL Pipeline


ETL means Extract, Transform, Load:
1. Extract data from its source.
2. Transform it into the required structure.
3. Load the cleaned result into a destination such as a database and CSV file.


F1


```python
def extract(file_path):
    extracted_df = pd.read_csv(file_path, parse_dates=["order_date"])
    print("Rows loaded:", len(extracted_df))
    return extracted_df

```

#F5. ETL vs ELT — sample explanation
In ETL, data is extracted, transformed before loading, and then stored in the destination. In ELT, raw data is extracted and loaded first, then transformed inside the destination system. This assessment implements ETL because the DataFrame is cleaned and transformed in Python before it is loaded into SQLite and a new CSV file
